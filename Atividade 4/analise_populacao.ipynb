# Análise de População 2010 x 2022
Leitura da tabela, agregação por estado e por município, e cálculo do crescimento populacional.

```python
import pandas as pd

df = pd.read_excel(
    "CD2022_Populacao_2010_Compatibilizada_20231222.xlsx",
    sheet_name="Municípios",
    header=2
)
df = df.dropna(how="all", axis=1)
df.columns = ["UF", "COD_UF", "COD_MUNIC", "MUNICIPIO", "POP_2010_SINOPSE", "POP_2010", "POP_2022"]
df.head()
```

## Agregado por estado (UF)

```python
pop_estado = (
    df.groupby("UF")[["POP_2010", "POP_2022"]]
    .sum()
    .reset_index()
)

pop_estado["CRESCIMENTO"] = pop_estado["POP_2022"] - pop_estado["POP_2010"]
pop_estado = pop_estado.sort_values("CRESCIMENTO", ascending=False).reset_index(drop=True)
pop_estado
```

```python
pop_estado.to_csv(r"pop_estado_crescimento.csv", sep=";", index=False)
```

## Agregado por município

```python
pop_municipio = (
    df.groupby(["UF", "MUNICIPIO"])[["POP_2010", "POP_2022"]]
    .sum()
    .reset_index()
)

pop_municipio["CRESCIMENTO"] = pop_municipio["POP_2022"] - pop_municipio["POP_2010"]
pop_municipio = pop_municipio.sort_values("CRESCIMENTO", ascending=False).reset_index(drop=True)
pop_municipio
```

```python
pop_municipio.to_csv(r"pop_municipio_crescimento.csv", sep=";", index=False)
```
