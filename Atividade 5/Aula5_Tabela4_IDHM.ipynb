# Atividade 5

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 1) Abrir o CSV

```python
path = r"Tabela4.csv"
df = pd.read_csv(path, sep=";", decimal=",", encoding="latin1", skiprows=1)
df
```

## 2) Limpar colunas vazias

```python
df = df.dropna(axis=1, how="all")
df
```

## 3) Ordene os estados por maior IDH em 2024

```python
df_2024 = df.sort_values(by="2024", ascending=False)
df_2024[["Sigla", "Estado", "2024"]]
```

## 4) Qual estado teve a maior melhora de IDH entre 2024 e 1991?

```python
df["melhora"] = df["2024"] - df["1991"]
df_melhora = df.sort_values(by="melhora", ascending=False)
df_melhora[["Sigla", "Estado", "1991", "2024", "melhora"]]
```

O estado foi Tocantins (TO), que saiu de 0,369 para 0,797, uma melhora de 0,428.

## 5) Existe algum estado em que o IDH piorou?

```python
piorou = df[df["melhora"] < 0]
piorou[["Sigla", "Estado", "1991", "2024", "melhora"]]
```

Nao existe nenhum estado em que o IDH piorou

## 6) Passar a tabela do formato largo para o formato longo (melt)

```python
anos = [c for c in df.columns if str(c).isdigit()]
print(anos)

id_vars = [c for c in df.columns if c not in anos]
df_longo = df.melt(
    id_vars=id_vars,
    value_vars=anos,
    var_name="Ano",
    value_name="IDH",
)
df_longo["Ano"] = df_longo["Ano"].astype(int)
df_longo["IDH"] = pd.to_numeric(df_longo["IDH"], errors="coerce")
df_longo.head()
```

## 7) Plote apenas Minas Gerais

```python
mg = df_longo[df_longo["Sigla"] == "MG"].sort_values("Ano")

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(mg["Ano"], mg["IDH"], marker="o", color="#1f77b4")

ax.set_title("Evolucao do IDH - Minas Gerais (1991-2024)")
ax.set_xlabel("Ano")
ax.set_ylabel("IDH")
fig.tight_layout()
plt.show()
```

## 8) Plotar a evolucao do IDH de cada estado

```python
fig, ax = plt.subplots(figsize=(12, 7))

for sigla, grupo in df_longo.groupby("Sigla"):
    grupo = grupo.sort_values("Ano")
    ax.plot(grupo["Ano"], grupo["IDH"], marker="o", markersize=3, linewidth=1.5, label=sigla)

ax.set_title("Evolucao do IDH por estado (1991-2024)")
ax.set_xlabel("Ano")
ax.set_ylabel("IDH")
ax.set_ylim(0.3, 0.9)
ax.legend(ncol=3, bbox_to_anchor=(1.02, 1), loc="upper left", title="UF")
fig.tight_layout()
plt.show()
```
